```python
class Turnoespera:
    def __init__(self , numero_turno, nombre_paciente, hora_llegada , motivo_consulta):
        self.numero_turno = numero_turno
        self.nombre_paciente = nombre_paciente
        self.hora_llegada = hora_llegada
        self.motivo_consulta = motivo_consulta
        self.sig = None
        self.ant = None

    def mostrar(self):
        print(f"Turno: {self.numero_turno} | Paciente: {self.nombre_paciente} | Hora: {self.hora_llegada} | Motivo: {self.motivo_consulta}")
```

```python
class Salaespera:
    def __init__(self ,):
        self.cabeza = None
        self.cola = None 
    
    def llegada_paciente(self , turno):
        if self.cabeza is None:
            self.cabeza = self.cola = turno
        else :
            self.cola.sig = turno
            turno.ant = self.cola 
            self.cola = turno

    def siguiente(self , numero_turno ):
        actual = self.buscar(numero_turno)
        if actual and actual.sig:
            return actual.sig
        return None 
    
    def anterior(self , numero_turno):
        actual = self.buscar(numero_turno)
        if actual and actual.ant:
            return actual.ant
        return None 
    
    def buscar(self, numero_turno):
        actual= self.cabeza
        while actual : 
            if actual.numero_turno == numero_turno : 
                return actual 
            actual = actual.sig
        return None 
    
    def emergencia(self, numero_turno):
            actual = self.cabeza
            anterior = None 
            while actual : 
                if actual.numero_turno == numero_turno:
                    if anterior:
                        anterior.sig = actual.sig 
                    else:
                        self.cabeza = actual.sig

                    actual.sig= self.cabeza
                    self.cabeza= actual
                    return None
                
                anterior= actual
                actual= actual.sig
    
    def mostrar_lista(self):
        actual = self.cabeza
        while actual:
            actual.mostrar()
            actual=actual.sig
```

```python
sala = Salaespera()

sala.llegada_paciente(Turnoespera(101, "Pedro", "9.00", "Dolor de cabeza"))
sala.llegada_paciente(Turnoespera(102, "Bryan", "10.00", "Fiebre"))
sala.llegada_paciente(Turnoespera(103, "Israel", "9.45", "Fractura de muñeca"))


print("Lista")
sala.mostrar_lista()


# Mostrar solo el siguiente del 101
print("\nSiguiente del turno 101:")
paciente = sala.siguiente(101)

if paciente:
    paciente.mostrar()

print("\nSiguiente del turno 102:")
paciente = sala.siguiente(102)

if paciente:
    paciente.mostrar()

print("\nEmergencia :")
sala.emergencia(103)

sala.mostrar_lista()
```

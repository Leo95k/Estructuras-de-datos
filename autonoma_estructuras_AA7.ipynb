```python
class GrafoEcoVuelo:
    def __init__(self):
        self.red_rutas={}

    def añadir_ciudad(self, ciudad):
        if ciudad not in self.red_rutas:
            self.red_rutas[ciudad] = {}
            print (f"Cuidad '{ciudad}' añadida con exito")
        else: 
            print (f"Error: La cuidad '{ciudad}' ya existe")

    def eliminar_ciudad(self, ciudad):
        if ciudad in self.red_rutas:
            del self.red_rutas[ciudad]

            for origen in self.red_rutas:
                if ciudad in self.red_rutas[origen]:
                    del self.red_rutas[origen][ciudad]

            print(f"Exito cuidad '{ciudad}' eliminada con exito")
        else:
            print("Error cuidad no encontrada")


    def añadir_ruta(self , origen , destino , distancia):
        if origen in self.red_rutas and destino in self.red_rutas:
            self.red_rutas[origen][destino] = distancia
            print(f"Ruta '{origen} -> {destino}' ({distancia}km) creada con exito")
        else:
            print ("Error ambas cuidades deben existir en el sistema para crear la ruta")


    def eliminar_ruta(self , origen , destino):
        if origen in self.red_rutas and destino in self.red_rutas[origen]:
            del self.red_rutas[origen][destino]
            print (f"Ruta '{origen} -> {destino}' eliminada con exito")
        else:
            print("Error la ruta no existe")

    def buscar_ruta(self, origen , destino):
        if origen in self.red_rutas and destino in self.red_rutas[origen]:
            print (f"Si existe un vuelo directo de '{origen}' a '{destino}'")
        else: 
            print(f"No hay vuelo directo de '{origen}' a '{destino}'")

    def buscar_ciudad(self, ciudad):
        if ciudad in self.red_rutas:
            print(f"La cuidad '{ciudad}' si esta registrada ")
        else:
            print(f"La cuidad '{ciudad}' no esta registrada")

    def destinos_directos(self , cuidad_origen):
        if cuidad_origen in self.red_rutas:
            destinos =list(self.red_rutas[cuidad_origen].keys())
            print (f"Destinos directos desde '{cuidad_origen}' : {destinos}")   
        else: 
            print("Error cuidad no encontrada")

    def consultar_distancia(self , origen , destino):
        if origen in self.red_rutas and destino in self.red_rutas[origen]:
            distancia = self.red_rutas[origen][destino]
            print(f"La distancia de '{origen}' a '{destino}' es de {distancia} km")
        else:
            print ("Error ruta no encontrada")

    def calcular_conectividad(self , ciudad):
        if ciudad not in self.red_rutas: 
            print("Error cuidad no encontrada")
            return

        vuelos_salida= len(self.red_rutas[ciudad])
        vuelos_ingreso= sum( 1 for origen in self.red_rutas if ciudad in self.red_rutas[origen])
        print (f"Conectividad de '{ciudad}' : '{vuelos_salida}' salidas directas , {vuelos_ingreso} ingresos directos")

    def recorrido_bfs(self , ciudad_inicio):
        if ciudad_inicio not in self.red_rutas:
            print("Error cuidad no existe")
            return 
        
        visitados = set([ciudad_inicio])
        cola =[ciudad_inicio]
        recorrido = []

        while cola: 
            actual =cola.pop(0)
            recorrido.append(actual)
            for vecino in self.red_rutas[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    cola.append(vecino)

        print(f"Recorrido BSF (anchura) desde '{ciudad_inicio}': {recorrido}")

    def recorrido_dfs(self , ciudad_inicio):
        if ciudad_inicio not in self.red_rutas:
            print("Error ciudad inicial no existe")
            return

        visitados = set()
        recorrido= []

        def dfs(ciudad):
            visitados.add(ciudad)
            recorrido.append(ciudad)

            for vecino in self.red_rutas[ciudad]:
                if vecino not in visitados:
                    dfs(vecino)

        dfs(ciudad_inicio)
        print(f"Recorrido DFS (profundidad) desde '{ciudad_inicio}': {recorrido}")  

        
```

```python
import tkinter as tk
from tkinter import ttk, messagebox, simpledialog
import sys


class RedireccionConsola:
    def __init__(self, text_widget):
        self.text_widget = text_widget

    def write(self, text):
        self.text_widget.insert(tk.END, text)
        self.text_widget.see(tk.END)

    def flush(self):
        pass

class InterfazEcoVuelo:
    def __init__(self, root, grafo):
        self.root = root
        self.grafo = grafo

        self.root.title("Sistema de Rutas Aéreas EcoVuelo")
        self.root.geometry("1000x720")
        self.root.configure(bg="#0f172a")

     
        header = tk.Frame(root, bg="#1e293b", height=90)
        header.pack(fill="x", side="top")
        tk.Label(header, text="SISTEMA DE RUTAS AÉREAS ECOVUELO", font=("Segoe UI", 20, "bold"), fg="#38bdf8", bg="#1e293b").pack(pady=(16, 0))
        tk.Label(header, font=("Segoe UI", 10), fg="#cbd5e1", bg="#1e293b").pack()

       
        cuerpo = tk.Frame(root, bg="#0f172a")
        cuerpo.pack(fill="both", expand=True, padx=15, pady=15)

        panel_botones = tk.Frame(cuerpo, bg="#0f172a")
        panel_botones.pack(side="left", fill="y", padx=(0, 15))

       
        panel_derecho = tk.Frame(cuerpo, bg="#0f172a")
        panel_derecho.pack(side="left", fill="both", expand=True)

        tk.Label(panel_derecho, text="Consola de Resultados", font=("Segoe UI", 11, "bold"), bg="#0f172a", fg="#38bdf8", anchor="w").pack(fill="x", pady=(0, 6))

        marco_texto = tk.Frame(panel_derecho, bg="#1e293b", bd=1, relief="solid")
        marco_texto.pack(fill="both", expand=True)

        scrollbar = tk.Scrollbar(marco_texto)
        scrollbar.pack(side="right", fill="y")

        self.texto = tk.Text(marco_texto, font=("Consolas", 11), bg="#0b1220", fg="#7dd3fc", insertbackground="#7dd3fc", bd=0, padx=12, pady=12, wrap="word", yscrollcommand=scrollbar.set)
        self.texto.pack(fill="both", expand=True)
        scrollbar.config(command=self.texto.yview)

       
        sys.stdout = RedireccionConsola(self.texto)
        print("Interfaz inicializada correctamente.")
        print("Bienvenido al Sistema. Utiliza los botones para interactuar.\n" + "-"*50 + "\n")

        grupos = [
            ("Ciudades", "#0ea5e9", [("Añadir Ciudad", self.btn_add_ciudad), ("Eliminar Ciudad", self.btn_del_ciudad), ("Buscar Ciudad", self.btn_buscar_ciudad)]),
            ("Rutas", "#22c55e", [("Añadir Ruta", self.btn_add_ruta), ("Eliminar Ruta", self.btn_del_ruta), ("Buscar Ruta", self.btn_buscar_ruta)]),
            ("Análisis", "#f59e0b", [("Destinos Directos", self.btn_destinos), ("Distancia", self.btn_distancia), ("Conectividad", self.btn_conectividad), ("BFS", self.btn_bfs), ("DFS", self.btn_dfs)]),
            ("Utilidades", "#a855f7", [("Limpiar Pantalla", self.limpiar)])
        ]

        for titulo, color, botones in grupos:
            marco = tk.LabelFrame(panel_botones, text=titulo, font=("Segoe UI", 10, "bold"), bg="#f1f5f9", fg="#1e293b", bd=0, padx=8, pady=8)
            marco.pack(fill="x", pady=(0, 10))
            for txt, cmd in botones:
                tk.Button(marco, text=txt, command=cmd, font=("Segoe UI", 10), bg=color, fg="white", relief="flat", width=20, cursor="hand2").pack(fill="x", pady=2)

    #Funcion agregada 
    def limpiar(self):
        self.texto.delete("1.0", tk.END)
        print("Pantalla limpiada.\n" + "-"*50 + "\n")

   
    def btn_add_ciudad(self):
        c = simpledialog.askstring("Añadir", "Nombre de la ciudad:")
        if c: 
            self.grafo.añadir_ciudad(c.strip())
            print("-" * 50)

    def btn_del_ciudad(self):
        c = simpledialog.askstring("Eliminar", "Nombre de la ciudad:")
        if c: 
            self.grafo.eliminar_ciudad(c.strip())
            print("-" * 50)

    def btn_add_ruta(self):
        o = simpledialog.askstring("Ruta", "Origen:")
        if not o: return
        d = simpledialog.askstring("Ruta", "Destino:")
        if not d: return
        km = simpledialog.askfloat("Ruta", "Distancia:")
        if km is not None:
            self.grafo.añadir_ruta(o.strip(), d.strip(), km)
            print("-" * 50)

    def btn_del_ruta(self):
        o = simpledialog.askstring("Eliminar Ruta", "Origen:")
        if not o: return
        d = simpledialog.askstring("Eliminar Ruta", "Destino:")
        if d:
            self.grafo.eliminar_ruta(o.strip(), d.strip())
            print("-" * 50)

    def btn_buscar_ciudad(self):
        c = simpledialog.askstring("Buscar", "Ciudad:")
        if c: 
            self.grafo.buscar_ciudad(c.strip())
            print("-" * 50)

    def btn_buscar_ruta(self):
        o = simpledialog.askstring("Buscar Ruta", "Origen:")
        if not o: return
        d = simpledialog.askstring("Buscar Ruta", "Destino:")
        if d: 
            self.grafo.buscar_ruta(o.strip(), d.strip())
            print("-" * 50)

    def btn_destinos(self):
        c = simpledialog.askstring("Destinos", "Ciudad origen:")
        if c: 
            self.grafo.destinos_directos(c.strip())
            print("-" * 50)

    def btn_distancia(self):
        o = simpledialog.askstring("Distancia", "Origen:")
        if not o: return
        d = simpledialog.askstring("Distancia", "Destino:")
        if d: 
            self.grafo.consultar_distancia(o.strip(), d.strip())
            print("-" * 50)

    def btn_conectividad(self):
        c = simpledialog.askstring("Conectividad", "Ciudad:")
        if c: 
            self.grafo.calcular_conectividad(c.strip())
            print("-" * 50)

    def btn_bfs(self):
        c = simpledialog.askstring("BFS", "Ciudad de inicio:")
        if c: 
            self.grafo.recorrido_bfs(c.strip())
            print("-" * 50)

    def btn_dfs(self):
        c = simpledialog.askstring("DFS", "Ciudad de inicio:")
        if c: 
            self.grafo.recorrido_dfs(c.strip())
            print("-" * 50)



if __name__ == "__main__":
    root = tk.Tk()
    sistema = GrafoEcoVuelo() 

    sistema.red_rutas = {
        "Guayaquil": {"Manta": 190.0, "Machala": 185.0, "Salinas": 140.0},
        "Manta": {"Guayaquil": 190.0, "Portoviejo": 40.0, "Esmeraldas": 350.0},
        "Machala": {"Guayaquil": 185.0},
        "Salinas": {"Guayaquil": 140.0},
        "Portoviejo": {"Manta": 40.0},
        "Esmeraldas": {"Manta": 350.0}
    }
    app = InterfazEcoVuelo(root, sistema)

    print("REGISTRANDO RED DE LA COSTA")
    for ciudad, destinos in sistema.red_rutas.items():
        print(f"Ciudad: {ciudad} -> Conexiones: {destinos}")
    print("\n" + "="*40 + "\n")
    
    def on_closing():
        sys.stdout = sys.__stdout__
        root.destroy()
        
    root.protocol("WM_DELETE_WINDOW", on_closing)
    root.mainloop()
```

EJERCICIO 1: Historial de Atenciones (Lista Enlazada Simple)

Escenario: Registro cronológico de pacientes que ya finalizaron su consulta.

• Clase AtencionMedica:

Atributos: id_registro, nombre_paciente, fecha (DD/MM/AAAA), monto_consulta.

• La Estructura a utilizar: Una Lista Simple donde los nuevos registros se insertan siempre

al final.

• Operaciones Obligatorias:

1. Insertar registro.

2. Mostrar historial completo (recorrido simple).

• Parte Personalizada: Implementar un método filtrar_por_monto(valor_minimo) que

muestre solo las consultas cuyo costo sea mayor al valor ingresado por teclado.

```python
#Ceamos la clase AtencionMedica con sus atributos id_registro, nombre_paciente, fecha y monto_consulta
class AtencionMedica:
    def __init__(self,id_registro,nombre_paciente,fecha,monto_consulta):
        self.id_registro = id_registro
        self.nombre_paciente = nombre_paciente
        self.fecha = fecha
        self.monto_consulta = monto_consulta
```

```python

#Creamos la clase Paciente con su atributo de atencion medica

class Paciente:
    def __init__(self,atencion_medica):
        self.atencion_medica = atencion_medica
        self.sig = None
```

```python
#Creamos ahurita si la clase de listasimple

class ListaSimple:
    def __init__(self):
        self.cabeza = None

    def insertar(self,atencion_medica):
        nuevo = Paciente(atencion_medica)

        if self.cabeza is None:
            self.cabeza = nuevo
        else:
            actual = self.cabeza
            while actual.sig is not None:
                actual = actual.sig
            actual.sig = nuevo

    def mostrar(self):
        actual = self.cabeza 
        while actual is not None:
            paciente = actual.atencion_medica
            
            print ("iD:", paciente.id_registro)
            print ("Paciente:", paciente.nombre_paciente)
            print ("Fecha:", paciente.fecha)
            print ("Monto de consulta:", paciente.monto_consulta)

            actual = actual.sig

    def filtrar_por_monto(self, valor_minimo):
        actual = self.cabeza
        while actual is not None:
            paciente = actual.atencion_medica
            if paciente.monto_consulta > valor_minimo:
                print("Paciente:", paciente.nombre_paciente)
                print("Monto:", paciente.monto_consulta)
            actual = actual.sig
```

```python
list = ListaSimple()
```

```python
cantidad = int(input("Ingrese cantidad de registros: "))

for i in range(cantidad):

    print("\n===== REGISTRO", i + 1, "=====")

    id_registro = int(input("Ingrese ID: "))

    nombre = input("Ingrese nombre del paciente: ")

    fecha = input("Ingrese fecha (DD/MM/AAAA): ")

    monto = float(input("Ingrese monto de consulta: "))

    atencion = AtencionMedica(
        id_registro,
        nombre,
        fecha,
        monto
    )

    list.insertar(atencion)
```

```python
print("\nHistorial del Paciente")

list.mostrar()
```

```python
valor = float(input("\nIngrese monto mínimo: "))

print("\n Consulta filtrada de los pacientes")

list.filtrar_por_monto(valor)
```

```python
class MonitoreoPaciente:

    def __init__(self, id_cama, nombre_paciente, frecuencia_cardiaca, estado_alerta):

        self.id_cama = id_cama
        self.nombre_paciente = nombre_paciente
        self.frecuencia_cardiaca = frecuencia_cardiaca
        self.estado_alerta = estado_alerta
```

```python
class Guardado:

    def __init__(self, paciente):

        self.paciente = paciente

        self.sig = None
        self.ant = None
```

```python
class ListaDobleCircular:

    def __init__(self):

        self.cabeza = None

    def conectar_paciente(self, paciente):

        nuevo = Guardado(paciente)

        # Lista vacía
        if self.cabeza is None:

            self.cabeza = nuevo

            nuevo.sig = nuevo
            nuevo.ant = nuevo

        else:

            ultimo = self.cabeza.ant

            ultimo.sig = nuevo
            nuevo.ant = ultimo

            nuevo.sig = self.cabeza
            self.cabeza.ant = nuevo

        print("Paciente conectado correctamente")

    def mostrar_monitoreo(self):

        if self.cabeza is None:

            print("No hay pacientes en monitoreo")
            return

        actual = self.cabeza

        while True:

            p = actual.paciente

            print("Cama:", p.id_cama)
            print("Paciente:", p.nombre_paciente)
            print("Frecuencia Cardíaca:",
                  p.frecuencia_cardiaca)
            print("Estado:", p.estado_alerta)

            actual = actual.sig

            if actual == self.cabeza:
                break

    def desconectar_paciente(self, id_cama):
        if self.cabeza is None:
            print("La lista está vacía")
            return

        actual = self.cabeza

        while True:
            if actual.paciente.id_cama == id_cama:
                if actual.sig == actual:
                    self.cabeza = None
                
                else:
                    actual.ant.sig = actual.sig
                    actual.sig.ant = actual.ant
                    
                    if actual == self.cabeza:
                        self.cabeza = actual.sig
                
                print("Paciente desconectado")
                return

            # Avanzamos al siguiente nodo
            actual = actual.sig
            
            # Si volvimos al inicio, el paciente no existe
            if actual == self.cabeza:
                break

        print("Paciente no encontrado")

    def mostrar_criticos(self):

        if self.cabeza is None:

            print("No hay pacientes")
            return

        actual = self.cabeza

        print("\nPACIENTES CRÍTICOS")

        while True:

            p = actual.paciente

            if p.estado_alerta.lower() == "crítico":

                print("Paciente:",
                      p.nombre_paciente)
                print("Cama:", p.id_cama)

            actual = actual.sig

            if actual == self.cabeza:
                break
```

```python
uci = ListaDobleCircular()
```

```python
cantidad = int(input("Ingrese cantidad de pacientes: "))

for i in range(cantidad):

    print("\n===== PACIENTE", i + 1, "=====")

    id_cama = int(input("Ingrese ID de cama: "))

    nombre = input("Ingrese nombre del paciente: ")

    frecuencia = int(input("Ingrese frecuencia cardíaca: "))

    estado = input("Ingrese estado alerta: ")

    paciente = MonitoreoPaciente(
        id_cama,
        nombre,
        frecuencia,
        estado
    )

    uci.conectar_paciente(paciente)
```

```python
print("\n MONITOREO UCI ")

uci.mostrar_monitoreo()
```

```python
id_eliminar = int(input("\nIngrese ID cama a desconectar: "))

uci.desconectar_paciente(id_eliminar)
```

```python
print("\n MONITOREO ACTUALIZADO ")

uci.mostrar_monitoreo()
```

```python
uci.mostrar_criticos()
```

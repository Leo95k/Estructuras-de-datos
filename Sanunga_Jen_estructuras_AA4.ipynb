```python
class NodoPedido:
 def __init__(self, id_pedido: int, tipo: str, descripcion: str):

  self.id_pedido= id_pedido #Ejemplo: 501 , 502
  self.tipo = tipo #Unicamente "Estandar " o "Reclamo"
  self.descripcion = descripcion #Ejemplo: Hamburgesa con papas
  self.siguiente = None  #Puntero al siguiente nodo en memoria 

class GestorPedidos :

  def __init__ (self):
    # Punteros para la Cola de Pedidos Estándar (FIFO)

    self.cola_frente = None

    self.cola_fin = None
    # Puntero para la Pila de Reclamos Urgentes (LIFO)

    self.pila_tope = None

  def registrar_pedido(self , id_pedido: int , tipo: str , descripcion: str) -> None:

     nuevo = NodoPedido(id_pedido , tipo , descripcion)
     tipo_normalizado = tipo.strip() 

     if tipo_normalizado == "Estandar" :
       if self.cola_frente is None :
        self.cola_frente = nuevo
        self.cola_fin = nuevo
       else :
        self.cola_fin.siguiente = nuevo 
        self.cola_fin = nuevo 
        print(f"Registrando pedido: id {id_pedido} , Tipo : {tipo} , Descripcion : {descripcion}")        

     elif  tipo_normalizado =="Reclamo": 
      nuevo.siguiente= self.pila_tope
      self.pila_tope = nuevo
      print(f"Registrando pedido: id {id_pedido} , Tipo : {tipo} , Descripcion : {descripcion}")


  def despachar_siguiente(self) -> NodoPedido :

     if self.pila_tope is not None : 
      pedido = self.pila_tope
      self.pila_tope = self.pila_tope.siguiente
      return pedido

     elif  self.cola_frente is not None: 
      pedido = self.cola_frente
      self.cola_frente = self.cola_frente.siguiente
      if self.cola_frente is None : 
       self.cola_fin= None
      return pedido

     return None 

  def activar_contingencia_masiva(self) -> None :

    actual = self.cola_frente

    while actual is not None:  
      nodo_mover = actual 
      actual = actual.siguiente 

      nodo_mover.siguiente = self.pila_tope 
      self.pila_tope = nodo_mover

    self.cola_frente = None 
    self.cola_fin = None    
```

```python
gestor  = GestorPedidos() 
print("1.- Registro de pedido")
gestor.registrar_pedido(501 , "Estandar" , "Pizza " )
gestor.registrar_pedido(502, "Estandar " , "Chaulafan")
gestor.registrar_pedido(503 , "Reclamo" , "Falta de utencilios en el anterior pedido")
gestor.registrar_pedido(504 , "Estandar " , "Burritos") 


print ("2.- Despachando pedidos")
p= gestor.despachar_siguiente()
print(f"Despachar pedido : id {p.id_pedido} , Tipo: {p.tipo} , Descripcion :{p.descripcion}")

print ("3.- Activar modo contingencia masiva")
gestor.activar_contingencia_masiva()
print ("Cola congelada tranfiriendo a la pila de reclamos")

print ("4.- Despachando la contingencia")

while True:
    p= gestor.despachar_siguiente()
    if p is None :
        print("No quedan mas pedidos")
        break
    else:
        print(f"Despachar pedido:   id {p.id_pedido} , Tipo: {p.tipo} , Descripcion :{p.descripcion}" )
```
